```python
from sklearn.datasets import load_breast_cancer
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split

from sap_rpt_oss import SAP_RPT_OSS_Classifier
import random
import numpy as np
from semantic_dataset import read_carte_classification_datasets


dataset = read_carte_classification_datasets(max_samples=4000)
acc_list = []

for _dataset in dataset:
    X, y = _dataset
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5, random_state=42)

    clf = SAP_RPT_OSS_Classifier(max_context_size=8192, bagging=1)

    # X_train, y_train = X_train.iloc[:1], y_train.iloc[:1]
    
    clf.fit(X_train, y_train)

    prediction_probabilities = clf.predict_proba(X_test)

    predictions = clf.predict(X_test)
    # print("Accuracy", accuracy_score(y_test, predictions))
    acc_list.append(accuracy_score(y_test, predictions))
    break

print(acc_list)
# [0.507, 0.507, 0.507, 0.507, 0.507, 0.507, 0.5522875816993464, 0.507, 0.5176600441501104, 0.507, 0.5425149700598803]
```
